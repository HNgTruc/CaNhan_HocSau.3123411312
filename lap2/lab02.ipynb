## Bài 1. Tạo Tensor cơ bản
Tạo một Tensor số nguyên kích thước 2 × 3 và hiển thị kết quả.

```python
import torch

a = torch.tensor(
    [[1, 2, 3],
     [4, 5, 6]],
    dtype=torch.int32
)

print(a)
```

```python
# Yêu cầu: Cho biết kích thước, số chiều và kiểu dữ liệu của Tensor a.
print("Kich thuoc (shape):", a.shape)
print("So chieu (ndim):", a.ndim)
print("Kieu du lieu (dtype):", a.dtype)
```

**Nhận xét:** Tensor `a` có kích thước `torch.Size([2, 3])` (2 dòng, 3 cột), số chiều `ndim = 2` (ma trận 2 chiều), kiểu dữ liệu là `torch.int32` vì được khai báo tường minh `dtype=torch.int32`.

## Bài 2. Tạo các Tensor thường dùng
Thực hành tạo Tensor ngẫu nhiên, Tensor toàn số 0, toàn số 1 và ma trận đơn vị.

```python
import torch

a = torch.rand(3, 4)
b = torch.randn(3, 4)
c = torch.zeros(2, 3)
d = torch.ones(2, 3)
e = torch.eye(4)

print(a)
print(b)
print(c)
print(d)
print(e)
```

## Bài 3. Kiểm tra thông tin Tensor
Sử dụng shape, size(), ndim và dtype để kiểm tra các thuộc tính của Tensor.

```python
import torch

a = torch.zeros(2, 3, 4)

print("Tensor:")
print(a)
print("Shape:", a.shape)
print("Size:", a.size())
print("Number of dimensions:", a.ndim)
print("Data type:", a.dtype)
```

## Bài 4. Slicing và Indexing
Truy xuất một phần dữ liệu từ Tensor nhiều chiều.

```python
import torch

torch.manual_seed(0)
a = torch.randn(3, 4, 5)

print("Original tensor:")
print(a)

print("\nTensor a[1]:")
print(a[1])

print("\nTensor a[1:, 2:4]:")
print(a[1:, 2:4])
```

```python
# Yêu cầu:
# - Cho biết shape của a[1] và a[1:, 2:4]
# - Thử lấy hai phần tử cuối cùng trên chiều thứ ba
print("Shape a[1]:", a[1].shape)
print("Shape a[1:, 2:4]:", a[1:, 2:4].shape)

print("\nHai phan tu cuoi tren chieu thu ba:")
print(a[:, :, -2:])
```

**Nhận xét:** `a[1]` bỏ đi chiều đầu tiên (chỉ lấy khối thứ 2) nên shape còn `(4, 5)`. `a[1:, 2:4]` giữ nguyên 3 chiều nhưng chiều 0 chỉ còn 2 khối (từ index 1 trở đi) và chiều 1 chỉ còn 2 dòng (index 2, 3), nên shape là `(2, 2, 5)`.

## Bài 5. Reshape và Flatten
Thay đổi hình dạng Tensor mà không làm thay đổi số phần tử.

```python
import torch

torch.manual_seed(0)
a = torch.randn(3, 4)

print("Original:")
print(a)

print("\nFlatten:")
print(a.ravel())

print("\nReshape to 3 x 2 x 2:")
print(a.reshape(3, 2, 2))
```

```python
# Yêu cầu:
# - Kiểm tra số phần tử trước và sau khi reshape
# - Thử reshape Tensor thành kích thước 2 x 6
print("So phan tu truoc reshape:", a.numel())
print("So phan tu sau reshape (3,2,2):", a.reshape(3, 2, 2).numel())

b = a.reshape(2, 6)
print("\nReshape thanh 2 x 6:")
print(b)
```

**Nhận xét:** Reshape chỉ thay đổi cách "nhìn" dữ liệu (số dòng/cột/chiều), không làm thay đổi tổng số phần tử — luôn phải giữ đúng tích các chiều bằng số phần tử ban đầu (ở đây là 12).

## Bài 6. Phép toán và nhân ma trận
Thực hiện các phép toán phần tử và phép nhân ma trận bằng toán tử @.

```python
import torch

torch.manual_seed(0)
a = torch.randn(2, 3)
b = torch.randn(2, 3)

print("a:")
print(a)
print("b:")
print(b)

print("\na + b:")
print(a + b)

print("\na / b:")
print(a / b)

print("\na^2:")
print(a ** 2)

print("\nMatrix multiplication:")
print(a @ b.T)
```

```python
# Yêu cầu: cho biết shape của a @ b.T
print("Shape a @ b.T:", (a @ b.T).shape)
```

**Nhận xét:** `a * b` là phép nhân **theo từng phần tử tương ứng** (element-wise), yêu cầu 2 Tensor cùng shape, kết quả cùng shape (2, 3). `a @ b.T` là phép **nhân ma trận** thực sự (dot product theo hàng-cột), ở đây `a` có shape (2,3), `b.T` có shape (3,2), nên kết quả có shape (2, 2).

## Bài 7. Các hàm thống kê trên Tensor
Tính trung bình, độ lệch chuẩn và tổng tích lũy theo từng cột.

```python
import torch

torch.manual_seed(0)
a = torch.randn(3, 4)

print(a)

print("\nMean (dim=0):")
print(torch.mean(a, dim=0))

print("\nStandard deviation (dim=0):")
print(torch.std(a, dim=0))

print("\nCumulative sum (dim=0):")
print(torch.cumsum(a, dim=0))
```

```python
# Yêu cầu: thay dim=0 bằng dim=1
print("Mean (dim=1):")
print(torch.mean(a, dim=1))

print("\nStandard deviation (dim=1):")
print(torch.std(a, dim=1))

print("\nCumulative sum (dim=1):")
print(torch.cumsum(a, dim=1))
```

**Nhận xét:** `dim=0` tính toán **dọc theo cột** (kết quả có 4 giá trị — mỗi giá trị đại diện 1 cột), còn `dim=1` tính toán **dọc theo dòng** (kết quả có 3 giá trị — mỗi giá trị đại diện 1 dòng). Nói cách khác, `dim` chỉ định chiều bị "gộp lại" khi tính toán.

## Bài 8. Autograd cơ bản
Tạo biến có requires_grad=True và dùng backward() để tính đạo hàm tự động.

```python
import torch

x = torch.tensor(
    3.6,
    requires_grad=True
)

y = x * x
y.backward()

print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

```python
# Yêu cầu 1: Tính thủ công đạo hàm của y = x^2 tại x = 3.6 và so sánh với x.grad
# dy/dx = 2x --> tai x=3.6: dy/dx = 7.2
dy_dx_thu_cong = 2 * 3.6
print("Dao ham thu cong dy/dx tai x=3.6:", dy_dx_thu_cong)
print("x.grad (PyTorch tinh):", x.grad.item())
print("So sanh: trung khop!" if abs(dy_dx_thu_cong - x.grad.item()) < 1e-6 else "Khac nhau")
```

```python
# Yêu cầu 2: Đổi hàm thành y = 3*x**2 + 2*x + 1 và kiểm tra gradient
x2 = torch.tensor(3.6, requires_grad=True)
y2 = 3 * x2 ** 2 + 2 * x2 + 1
y2.backward()

# dao ham thu cong: dy/dx = 6x + 2 --> tai x=3.6: 6*3.6+2 = 23.6
dy_dx_thu_cong2 = 6 * 3.6 + 2
print("y2 =", y2)
print("x2.grad (PyTorch tinh):", x2.grad.item())
print("Dao ham thu cong:", dy_dx_thu_cong2)
```

**Nhận xét:** Với `y = x²`, đạo hàm `dy/dx = 2x`, tại `x = 3.6` cho kết quả `7.2` — khớp hoàn toàn với `x.grad` mà PyTorch tự tính qua `backward()`. Với `y = 3x² + 2x + 1`, đạo hàm `dy/dx = 6x + 2`, tại `x = 3.6` cho `23.6` — cũng khớp với `x2.grad`. Điều này cho thấy autograd tính đạo hàm chính xác theo quy tắc giải tích, không phải xấp xỉ.

## Bài 9. Gradient Descent tìm hệ số đa thức
Sinh dữ liệu từ đa thức y = x² + 2x + 3, sau đó dùng NAdam và autograd để tìm lại ba hệ số.

```python
import numpy as np
import torch

torch.manual_seed(0)
np.random.seed(0)

polynomial = np.poly1d([1, 2, 3])
N = 20

X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("Learned coefficients:")
print(w)
```

**Nhận xét:**
- Ba hệ số học được (`w`) sau 1000 vòng lặp xấp xỉ rất gần với `[1, 2, 3]` — đúng với đa thức gốc `y = x² + 2x + 3`, chứng tỏ Gradient Descent đã hội tụ đúng.
- `optimizer.zero_grad()`: xóa gradient tích lũy từ vòng lặp trước (PyTorch mặc định cộng dồn gradient, nếu không xóa sẽ tính sai).
- `mse.backward()`: tính gradient của hàm mất mát `mse` theo từng tham số `w` bằng autograd (lan truyền ngược).
- `optimizer.step()`: cập nhật giá trị `w` theo hướng ngược gradient (giảm dần sai số), theo công thức của thuật toán NAdam.

## Bài 10. Giải hệ phương trình bằng Autograd
Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình.

```python
import random
import torch

random.seed(0)
torch.manual_seed(0)

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2

    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4

    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

```python
# Yêu cầu: Kiểm tra nghiệm thu được với 4 phương trình
print("A + B =", (A + B).item(), " (ky vong = 9)")
print("C - D =", (C - D).item(), " (ky vong = 1)")
print("A + C =", (A + C).item(), " (ky vong = 8)")
print("B - D =", (B - D).item(), " (ky vong = 2)")
```

**Nhận xét:** Cả 4 giá trị kiểm tra đều xấp xỉ đúng vế phải (9, 1, 8, 2), cho thấy nghiệm tìm được thỏa hệ phương trình. Tuy nhiên bài toán có **4 ẩn nhưng chỉ 4 phương trình độc lập tuyến tính** — thực chất hệ này có vô số nghiệm (thiếu 1 điều kiện để xác định duy nhất D, các biến còn lại phụ thuộc vào D), vì nếu đặt D = d bất kỳ thì A, B, C đều có thể suy ra tương ứng mà vẫn thỏa cả 4 phương trình. Vì vậy, giá trị A, B, C, D cụ thể tìm được phụ thuộc vào điểm khởi tạo ngẫu nhiên ban đầu.

### d) Thay đa thức trong Bài 9 bằng y = 2x² - 3x + 5; huấn luyện và cho biết các hệ số tìm được

```python
import numpy as np
import torch

torch.manual_seed(0)
np.random.seed(0)

polynomial = np.poly1d([2, -3, 5])
N = 20

X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(2000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("Learned coefficients (ky vong ~ [2, -3, 5]):")
print(w)
```

**Nhận xét:** Sau khi huấn luyện, ba hệ số học được xấp xỉ rất gần `[2, -3, 5]` — đúng với đa thức gốc `y = 2x² - 3x + 5`, cho thấy phương pháp Gradient Descent với optimizer NAdam áp dụng được cho bất kỳ đa thức bậc 2 nào, chỉ cần đổi hệ số sinh dữ liệu ban đầu.
